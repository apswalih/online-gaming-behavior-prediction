## **Saved Model Inference Demo**


**This notebook demonstrates:**

1. Loading the trained pipeline model
2. Loading label mappings
3. Preparing new player data
4. Running predictions
5. Decoding predictions into readable labels
6. Viewing prediction confidence scores


### **Import Libraries**

```python
import joblib
import pandas as pd
from pathlib import Path
```

### **Project Path Setup**

```python
# Get project root directory safely

BASE_DIR = Path().resolve().parent

# If notebook is inside /notebooks folder
if BASE_DIR.name == "notebooks":
    BASE_DIR = BASE_DIR.parent

# Models directory
MODELS_DIR = BASE_DIR / "models"
```

### **Load Trained Model**

```python
MODEL_PATH = MODELS_DIR / "xgboost_pipeline.joblib"

model = joblib.load(MODEL_PATH)

print("Model loaded successfully")
```

### **Load Label Mapping**

```python
LABEL_MAPPING_PATH = MODELS_DIR / "label_mapping.joblib"

label_mapping = joblib.load(LABEL_MAPPING_PATH)

print("Label mapping loaded successfully")
```

### **Create Reverse Mapping**

**Converts:**

- 0 -> Low
- 1 -> Medium
- 2 -> High

```python
reverse_mapping = {v: k for k, v in label_mapping.items()}

print("Reverse mapping created")
print(reverse_mapping)
```

### **New Player Data**

```python
# Example player information for prediction

new_data = pd.DataFrame([{
    "Age": 25,
    "PlayTimeHours": 120,
    "SessionsPerWeek": 7,
    "AvgSessionDurationMinutes": 60,
    "PlayerLevel": 15,
    "AchievementsUnlocked": 10,
    "InGamePurchases": 0,
    "Gender": "Male",
    "Location": "USA",
    "GameGenre": "RPG",
    "GameDifficulty": "Hard"
}])
```

### **Model Prediction**

```python
prediction = model.predict(new_data)

# Extract numeric prediction
predicted_class = prediction[0]


# CONVERT PREDICTION TO READABLE LABEL
final_prediction = reverse_mapping[predicted_class]

print("Predicted Engagement Level:", final_prediction)
```

### **Prediction Probabilities** 

```python
probability_scores = model.predict_proba(new_data)[0]

print("Prediction Probability Scores:")

for class_id, probability in enumerate(probability_scores):
    
    class_name = reverse_mapping[class_id]
    
    print(f"{class_name}: {probability:.4f}")
```

### **Model Information**

```python
print("Model Expected Features:")
print(model.feature_names_in_)
```

```python
print("Model Classes:")
print(model.classes_)
```

### **Final Summary**

```python
print("=========== FINAL RESULT ===========")
print(f"Predicted Engagement Level : {final_prediction}")

highest_probability = max(probability_scores)
print(f"Prediction Confidence      : {highest_probability:.2%}")
```
